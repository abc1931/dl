```python
import tensorflow as tf
from tensorflow.keras.datasets import cifar10
from tensorflow.keras.applications import VGG16
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Flatten, Dense
from tensorflow.keras.utils import to_categorical

# 1. Load CIFAR-10 dataset
(itrain, ltrain), (itest, ltest) = cifar10.load_data()

# 2. Preprocess the data
itrain = itrain.astype("float32") / 255.0
itest = itest.astype("float32") / 255.0
ltrain = to_categorical(ltrain, 10)
ltest = to_categorical(ltest, 10)

# 3. Load pre-trained VGG16 model
basem = VGG16(
    weights="imagenet",
    include_top=False,
    input_shape=(32, 32, 3)
)

# 4. Freeze the pre-trained layers
for layer in basem.layers:
    layer.trainable = False

# 5. Create a new model
semodel = Sequential([
    basem,
    Flatten(),
    Dense(256, activation="relu"),
    Dense(10, activation="softmax")
])

# 6. Compile the model
semodel.compile(
    optimizer="adam",
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

# 7. Train the model
semodel.fit(
    itrain,
    ltrain,
    epochs=10,
    batch_size=32,
    validation_data=(itest, ltest)
)

# 8. Evaluate the model
test_loss, test_accuracy = semodel.evaluate(itest, ltest)
print("Test loss:", test_loss)
print("Test accuracy:", test_accuracy)
```
